# Description of Dataset
The data containes GBIF occurrence records and additional climate and environmental data for arthropod prey of selected species in California drylands as identified in Lortie et al. (2023).

## Reference:
Rachel King, Jenna Braun, Michael Westphal, & CJ Lortie. (2023). Compiled occurrence records for prey items of listed species found in California drylands with associated environmental data. Knowledge Network for Biocomplexity. doi:10.5063/F1VM49RH.

## 7. Subsetting
Complete the following subsetting tasks.
1. Select only the preserved specimens from prey.
2. Select observations between 1992 and 2000.
3. Select all rows for columns species and eventDate.
4. Select observations that are greater than 37° latitude and less than 39° latitude.

```python
# Select only the preserved specimens from prey
prey.basisOfRecord
preserved = prey[prey['basisOfRecord'] == "PRESERVED_SPECIMEN"]
preserved
```

```python
# Select observations between 1992 and 2000
years_92to00 = prey[(prey['year'] >= 1992) & (prey['year'] <= 2000)]
print(years_92to00['year'].value_counts())
years_by_np = prey[prey['year'].between(1992, 2000)] # Different np method, same results
print(years_by_np['year'].value_counts())
```

```python
# Select all rows for columns `species` and `eventDate`
prey[['species', 'eventDate']]
```

```python
# Select observations that are greater than 37° latitude and less than 39° latitude
prey.head()
prey_box = prey[(prey['lat'] > 37) & (prey['long'] < 39)]
prey_box[['lat', 'long']]
```
