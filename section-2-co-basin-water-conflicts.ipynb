## 8. Exploratory wrangl

```python
# Step 1 select state column where conflict present
conflict_states = col_riv_conflict.loc[col_riv_conflict['Conflict Present'] == 'Y', 'State']

# Step 2: Split on semicolon delimiter
states_split = conflict_states.str.split(';')

# Step 3: Explode lists into individual rows
states_explode = states_split.explode()

#Step 4: Unique values, keeping NAs
states_explode.unique()
```

```python
states_strip = states_explode.str.strip(' ').unique()
states_strip
```
