## 8. Exploratory wrangling

```python
# Step 1 select state column where conflict present
conflict_states = col_riv_conflict.loc[col_riv_conflict['Conflict Present'] == 'Y', 'State']

# Step 2: Split on semicolon delimiter
states_split = conflict_states.str.split(';')

# Step 3: Explode lists into individual rows
states_explode = states_split.explode()

#Step 4: Unique values, keeping NAs
states_explode.unique()
```

## 9. Find unique state codes

```python
states_strip = states_explode.str.strip(' ').unique()
states_strip
```

```python
states_conflict = (
    col_riv_conflict.loc[col_riv_conflict['Conflict Present'] == 'Y', 'State']
    .str.split(';')
    .explode()
    .str.strip(' ')
    .unique()
)
states_conflict
```

## 10. Visualize data exploration & discussion

```python
states_counts = (
    col_riv_conflict.loc[col_riv_conflict['Conflict Present'] == 'Y', 'State']
    .str.split(';')
    .explode()
    .str.strip(' ')
    .value_counts(dropna = False)
)
states_counts
```

```python
states_counts.plot(kind='bar',
              x='State', 
              title='Number of conflicts in states',
              xlabel='State',
              ylabel='Number of conflicts',
              color='purple'
              )
```
